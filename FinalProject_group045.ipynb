## Abstract

This study investigates whether California counties considered as college towns show different violent and property crime rates compared to other counties. Using the FBI’s NIBRS crime data from 2024, US Census data, and IPEDS college enrollment data. We analyzed crime rates across California counties and built classification models to test whether crime patterns could predict whether a county was considered a college town. Our findings showed weak correlations between crime rates and college towns, with all 3 models that we ran (Random Forest, KNN, Logistic Regression) performed near 50-50, which we believe suggests county-level crime rates alone are a poor predictor of whether it’s a college town or not.


To define college counties, we merged the IPEDs enrollment data with census population estimates to compute the student-to-population ratio for each county, where we then classified the top 25% as college-concentrated. To allow for fair comparisons across counties of different sizes, we joined FBI NIBRS records across agency, incident, and offense tables and normalized to offenses per 100,000 residents. Offense categories were grouped into either violent or property crime buckets, with larceny, theft, fraud, motor vehicle theft, and burglary counted for property crime, while assault, sex offenses, robbery, and homicide were counted for violent crime. 


Our three models were trained on the violent crime rate, the property crime rate, and the county population. We used the student-to-population ratio to define our target, and so we excluded it from the models to avoid data leakage. Our best-performing model was the Logistic Regression which achieved an ROC-AUC of 0.687 and accurately caught 77.4% of college-counties, but also diagnosed many false positives. The Random Forest and KNN both performed significantly worse with ROC-AUCs of .568 and .663, respectively. The Spearman correlation heatmap, geospatial heatmaps, and model results all point to the same conclusion, there is evidence of a weak correlation between crime rates and college town status. 


To determine whether or not there was a statistically significant difference between the property crime rate and the violent crime rate in college counties versus non-college counties, we performed a Mann-Whitney U test with our significance level threshold set at 0.025. For violent crime, we uncovered a statistically insignificant p-value of 0.1597, which led us to fail to reject the null-hypothesis, meaning there was no statistically significant difference between violent crime rates in college counties versus non-college counties. On the other hand, when we performed this test for property crime, we found that there was a statistically significant p-value of 0.005, meaning there was a statistically significant difference between property crime in college counties versus non-college counties.

## Background and Prior Work

Crime in college and university towns has been widely reported for the last few decades, with government reported statistics going back to the 1990s. The <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Jeanne Clery Disclosure of Campus Security Policy and Campus Crime Statistics Act requires that crime data for universities must be publicly reported through a Department of Education sponsored survey. There is an abundance of data on isolated college crimes, but less so on the crime rates of areas near universities. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)A study in the American Journal of Police compared campus and city crime rates, published in 1995 by Max L. Bromley from the University of South Florida, examined the relationship between the crime rates with the local communities that surround the university campus. The data used in the study relied heavily on the two largest universities in each state and on an average of 2 years for the population. In result, there was a similar widespread variation found when comparing the violent and property crime rates per 1,000 students or city residents. All of the comparisons concluded that the university's crime rates were significantly lower than those of their cities. In addition, there was a finding that property crime was the vast majority of crimes compared to violent crimes, and that the university property crimes were especially higher than those in their cities. Overall, the findings were consistent with the prior studies in the 1980s and the early 1990s. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)A research paper published on November 24 2020 by Kelsey Cundiff at University of Missouri-St. Louis analyzed campus proximity and neighborhood crime rates. It hypothesized with opportunity theory that neighborhoods within close proximity to college campuses will have higher crime rates because “college students are vulnerable to victimisation" due to convergence for their social lifestyles. It used data collected by the National Neighborhood Crime Survey, and analysis was done with a “multilevel negative binomial regression” to analyze the relationship between “proximity… to college campus[es] and rates of violent and property crime”. The researchers doing this study found that “bordering tracts higher rates of larceny, burglary, and robbery,” acting as proxies for “indicators of crime.” <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) A research study published on October 10, 2023 by Carter Prentice and his team conducted a study on whether there was a difference in homicide and suicide rates between college towns/cities vs non college towns. Prentice’s team conducted this study by analyzing county level mortality rates from the CDC death certificate data from the years 2015 to 2019. To determine if a county was deemed a college town they used the classifications of the American Communities Project (ACP) to determine whether it was a college town or not. In college towns, they found that there was a median 2.56 total homicides per 100000 population and in non college towns, a 3.11 total homicide median. But in college towns, they found that people aged less than 24 had a higher rate of firearm homicide compared to non college towns, showing elevated risk in younger people with firearms. The researchers found that the results for population level homicide were not significantly different from college towns and non college towns. They found that certain age groups had increased risk of homicide with no outliers in any counties.

There are still notable gaps in the existing research, since studies focus on the close proximity of crime near colleges, or on certain categories of crime(either property or violent crime). Additionally, these studies are fairly old, with the 2020 study at the University of Missouri St. Louis being the most recent to analyse campus proximity in relation to crime.  Additionally, the study conducted in 1995 was conducted for the South Florida region, so we hope to investigate the findings in a different region (California).

```python
# FBI NIBRS CLEANING

DATA_DIR = "data/00-raw/CA-2024"
OUT_DIR = "data/02-processed"
os.makedirs(OUT_DIR, exist_ok=True)

ag  = pd.read_csv(f"{DATA_DIR}/agencies.csv", encoding="latin1", low_memory=False)
off = pd.read_csv(f"{DATA_DIR}/NIBRS_OFFENSE.csv", low_memory=False)
ot  = pd.read_csv(f"{DATA_DIR}/NIBRS_OFFENSE_TYPE.csv", low_memory=False)
inc = pd.read_csv(f"{DATA_DIR}/NIBRS_incident.csv", low_memory=False)

# filter to CA + 2024
ag = ag[
    (ag["data_year"] == 2024) &
    (ag["state_postal_abbr"] == "CA") &
    (ag["publishable_flag"] == "Y")
]
off = off[off["data_year"] == 2024]
inc = inc[inc["data_year"] == 2024]

# county population
county_pop = (ag.groupby("county_name", as_index=False)["population"]
                .sum()
                .rename(columns={"population": "county_population"}))

# incident -> agency -> county
inc_ag = inc[["incident_id", "agency_id", "data_year"]].merge(
    ag[["agency_id", "data_year", "county_name"]],
    on=["agency_id", "data_year"],
    how="inner"
)

# offense -> offense name/category -> county
off_named = off.merge(
    ot[["offense_code", "offense_name", "offense_category_name"]],
    on="offense_code",
    how="left"
).merge(
    inc_ag[["incident_id", "county_name"]],
    on="incident_id",
    how="inner"
)

missing = off_named.isna().mean().sort_values(ascending=False)
display(missing.head(15))

# counts + rates
counts = (off_named.groupby(["county_name", "offense_category_name", "offense_name"], as_index=False)
                  .size()
                  .rename(columns={"size": "offense_rows"}))

final = counts.merge(county_pop, on="county_name", how="left")
final["offenses_per_100k"] = final["offense_rows"] / final["county_population"] * 100000

display(final.sort_values("offense_rows", ascending=False).head(5))

final.to_csv(f"{OUT_DIR}/ca2024_county_offense_rates.csv", index=False)
county_pop.to_csv(f"{OUT_DIR}/ca2024_county_population.csv", index=False)

print("Saved:",
      f"{OUT_DIR}/ca2024_county_offense_rates.csv",
      "and",
      f"{OUT_DIR}/ca2024_county_population.csv")
```

```python
# Load data
county_offense_data = pd.read_csv("./data/02-processed/ca2024_county_offense_rates.csv")

# Define crime categories
property_crimes = [
    "Larceny/Theft Offenses", "Fraud Offenses", "Motor Vehicle Theft",
    "Destruction/Damage/Vandalism of Property", "Burglary/Breaking & Entering",
    "Stolen Property Offenses", "Embezzlement", "Arson",
    "Counterfeiting/Forgery", "Gambling Offenses"
]

violent_crimes = [
    "Sex Offenses", "Assault Offenses", "Homicide Offenses",
    "Prostitution Offenses", "Sex Offenses, Non-forcible", "Weapon Law Violations",
    "Kidnapping/Abduction", "Robbery", "Animal Cruelty",
    "Pornography/Obscene Material", "Extortion/Blackmail",
    "Human Trafficking", "Bribery"
]

# Flag crime type
county_offense_data["violent"] = county_offense_data["offense_category_name"].isin(violent_crimes).astype(int)
county_offense_data["property"] = county_offense_data["offense_category_name"].isin(property_crimes).astype(int)

# Aggregate offenses per county
county_offenses_per_100k = (
    county_offense_data
    .assign(
        violent_offenses=lambda x: x["offenses_per_100k"] * x["violent"],
        property_offenses=lambda x: x["offenses_per_100k"] * x["property"]
    )
    .groupby("county_name")
    .agg(
        violent_per_100k=("violent_offenses", "sum"),
        property_per_100k=("property_offenses", "sum"),
        county_pop=("county_population", "mean")
    )
    .replace([np.inf, -np.inf], np.nan)
    .dropna()
)

county_offenses_per_100k.head(5)
```

#### Hypothesis Testing
We will perform two hypothesis test. One for violent crime and one for property crime. For both we will use Mann–Whitney U Test becasue it doesn't assume a normal distribution which our crime statistics aren't, and it works well with comparing different distributions.

Since we're doing two hypothesis test, we will use Bonferroni correction in order to account for our increased false positive risk.

Therefore:

α = (0.05) / (2 tests) = 0.025 will be our significance level threshold.

##### Violent Crime U Test

Null Hypothesis: 
The distribution of violent crime rates per 100,000 people is the same for college counties and non-college counties.

Alternative Hypothesis: 
The distribution of violent crime rates per 100,000 people differs between college counties and non-college counties.

```python
hypothesis_test_df = big_school_df.copy()

hypothesis_test_df["is_college_county"] = (hypothesis_test_df["EFTOTAL/county_pop"] >= hypothesis_test_df["EFTOTAL/county_pop"].quantile(0.75)).astype(int)

college_county_v = hypothesis_test_df[hypothesis_test_df["is_college_county"] == 1]["violent_per_100k"]
non_college_county_v = hypothesis_test_df[hypothesis_test_df["is_college_county"] == 0]["violent_per_100k"]

stat_v, p_v = mannwhitneyu(college_county_v, non_college_county_v, alternative="two-sided")
print("Violent crime p-value:", p_v) 
```

##### Distribution Plot

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=hypothesis_test_df[hypothesis_test_df["is_college_county"] == 1],
    x="violent_per_100k",
    label="College Counties",
    fill=True,
    alpha=0.4
)

sns.kdeplot(
    data=hypothesis_test_df[hypothesis_test_df["is_college_county"] == 0],
    x="violent_per_100k",
    label="Non-College Counties",
    fill=True,
    alpha=0.4
)

plt.title("Violent Crime Distribution")
plt.xlabel("Violent Crime per 100k")
plt.legend()
plt.show()
```

We fail to reject the null hypothesis. There was no statistically significant difference in violent crime rates observed between college and non-college counties.

#### Models
We create a binary target where 1 = "college-concentrated county" (top 25% enrollment share) and 0 = not college-concentrated. This is determined by whether a county's student-to-population ratio (EFTOTAL/county_pop) is at or above the 75th percentile. 617 school records are split into a training set of 493 (80%) and a test set of 124 (20%), with stratification to preserve the class imbalance (~75% negative, ~25% positive) in both sets.


```python
feature_cols = ["violent_per_100k", "property_per_100k", "county_pop", "EFTOTAL/county_pop"]
df_model = big_school_df[feature_cols].replace([np.inf, -np.inf], np.nan).dropna().copy()

# Proxy binary target: top quartile of enrollment share
y = (df_model["EFTOTAL/county_pop"] >= df_model["EFTOTAL/county_pop"].quantile(0.75)).astype(int)

# Avoid leakage by excluding ratio used to define y
X = df_model[["violent_per_100k", "property_per_100k", "county_pop"]]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train shape:", X_train.shape, "Test shape:", X_test.shape)
print("Class balance in y_train:")
print(y_train.value_counts(normalize=True))
```

##### Model 1: Random Forest

```python
rf_model = RandomForestClassifier(
    n_estimators=400, random_state=42, class_weight="balanced"
 )

rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

print("Random Forest ROC-AUC:", round(roc_auc_score(y_test, y_prob_rf), 4))
print(classification_report(y_test, y_pred_rf, digits=3))
```

This model achieved a ROC-AUC of 0.568 and an overall accuracy of 52.4%. It correctly identified only 41.9% of college-concentrated counties, suggesting crime rates and population have little predictive power in a tree model.

##### Model 2: KNN 


```python
knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", KNeighborsClassifier(n_neighbors=9))
])

knn_model.fit(X_train, y_train)
y_pred_knn = knn_model.predict(X_test)
y_prob_knn = knn_model.predict_proba(X_test)[:, 1]

print("KNN ROC-AUC:", round(roc_auc_score(y_test, y_prob_knn), 4))
print(classification_report(y_test, y_pred_knn, digits=3))
```

This model achieved a ROC-AUC of 0.668 and an overall accuracy of 71.8%, though could be misleading because it mostly just predicted "not a college town" for almost everything, and it only caught 9.7% of actual college-concentrated counties.

##### Model 3 Logistic Regression

```python
lr_model = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(max_iter=1000, random_state=42, class_weight="balanced"))
])

lr_model.fit(X_train, y_train)
y_pred_lr = lr_model.predict(X_test)
y_prob_lr = lr_model.predict_proba(X_test)[:, 1]

print("Logistic Regression ROC-AUC:", round(roc_auc_score(y_test, y_prob_lr), 4))
print(classification_report(y_test, y_pred_lr, digits=3))
```

This model achieved the best ROC-AUC of 0.687 and an accuracy of 60.5%, correctly identifying 77.4% of college-concentrated counties but with a lot of false positives. It still finds that crime rates alone are a weak predictor of college town status.
